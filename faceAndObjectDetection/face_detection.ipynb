```python

import cv2

cap = cv2.VideoCapture(0)

face_cascade = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")
eye_cascade = cv2.CascadeClassifier("haarcascade_eye_tree_eyeglasses.xml")
smile_cascade = cv2.CascadeClassifier("haarcascade_smile.xml")

while True:

    ret, frame = cap.read()

    if not ret:
        break

    boundingBox_end = ()

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(gray, 1.1, 5)

    # Top UI bar
    cv2.rectangle(
        frame,
        (0, 0),
        (640, 55),
        (25, 25, 25),
        -1
    )

    cv2.putText(
        frame,
        "VISION DETECTION",
        (20, 35),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (255, 255, 255),
        2
    )

    cv2.putText(
        frame,
        "LIVE",
        (535, 34),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (0, 255, 180),
        2
    )

    for (x, y, w, h) in faces:

        # Save face coordinates
        face_x = x
        face_y = y
        face_w = w
        face_h = h

        # Face circle
        cv2.circle(
            frame,
            (face_x + face_w // 2, face_y + face_h // 2),
            min(face_w, face_h) // 2,
            (255, 255, 0),
            3
        )

        # Face label
        cv2.rectangle(
            frame,
            (face_x, face_y - 30),
            (face_x + 125, face_y),
            (25, 25, 25),
            -1
        )

        cv2.putText(
            frame,
            "FACE",
            (face_x + 10, face_y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (255, 255, 0),
            2
        )

        boundingBox_end = (face_x + face_w, face_y + face_h)

        # Face region
        face_region_gray = gray[
            face_y:face_y + face_h,
            face_x:face_x + face_w
        ]

        face_region_color = frame[
            face_y:face_y + face_h,
            face_x:face_x + face_w
        ]

        # -------------------------
        # EYE DETECTION
        # -------------------------

        eyes = eye_cascade.detectMultiScale(
            face_region_gray,
            1.1,
            20
        )

        for (x, y, w, h) in eyes:

            cv2.circle(
                face_region_color,
                (x + w // 2, y + h // 2),
                min(w, h) // 2,
                (255, 100, 0),
                2
            )

            cv2.putText(
                face_region_color,
                "EYE",
                (x, y - 8),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.4,
                (255, 100, 0),
                1
            )

        # -------------------------
        # SMILE DETECTION
        # -------------------------

        smiles = smile_cascade.detectMultiScale(
            face_region_gray,
            10,
            10
        )

        if len(smiles) > 0:

            # Smile label BELOW THE FACE
            smile_y = face_y + face_h + 15

            cv2.rectangle(
                frame,
                (face_x, smile_y),
                (face_x + 120, smile_y + 32),
                (25, 25, 25),
                -1
            )

            cv2.putText(
                frame,
                "SMILE :)",
                (face_x + 8, smile_y + 23),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (255, 100, 255),
                2
            )

    # -------------------------
    # STATUS PANEL
    # -------------------------

    cv2.rectangle(
        frame,
        (10, frame.shape[0] - 55),
        (230, frame.shape[0] - 10),
        (25, 25, 25),
        -1
    )

    cv2.putText(
        frame,
        f"Faces: {len(faces)}",
        (25, frame.shape[0] - 27),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (255, 255, 255),
        1
    )

    cv2.putText(
        frame,
        "Press Q to exit",
        (frame.shape[1] - 150, frame.shape[0] - 25),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.45,
        (180, 180, 180),
        1
    )

    cv2.imshow("WebCam", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

### Video Processing

```python
# Recording and saving video using OpenCV
import cv2

cap = cv2.VideoCapture(0)

frame_hieght = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))

codec = cv2.VideoWriter_fourcc(*'XVID')
recorder = cv2.VideoWriter(filename="recording.avi", fps=360, frameSize=(frame_width, frame_hieght),fourcc=codec)
while True:
    ret, frame = cap.read()
    if not ret:
        break
    recorder.write(frame)
    cv2.imshow("recording...", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
recorder.release()
cv2.destroyAllWindows()
```

```python
# Detect Strong edges using canny Detection
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read(cv2.IMREAD_GRAYSCALE)
    edges = cv2.Canny(frame, 50, 150)
    if not ret:
        break
    cv2.imshow("webcam", frame)
    edges = cv2.resize(edges, None, None, 0.5, 0.5)
    cv2.imshow("Edges", edges)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()

```

```python
# Threasholding(either black or white)...
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    ret, thresh_image = cv2.threshold(cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY), 120, 255, cv2.THRESH_BINARY)
    cv2.imshow(f"{ret}", thresh_image)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
# Bitwise Operations on images
import numpy as np
import cv2

img1 = np.zeros(shape=(500, 300), dtype="uint8")
img2 = np.zeros(shape=(500, 300), dtype="uint8")

h1, w1 = img1.shape
h2, w2 = img2.shape

img1 = cv2.circle(img1, (100, 150), h1 // 4, 255, -1)
img2 = cv2.rectangle(img2, (w2 // 4, h2 // 4), (3 * w2 // 4, 3 * h2 // 4), 255, -1)

cv2.imshow("Img-1", img1)
cv2.imshow("Img-2", img2)

comman_part = cv2.bitwise_and(img1, img2)
combined_part = cv2.bitwise_or(img1, img2)
remaining_part = cv2.bitwise_not(combined_part)
cv2.imshow("remaining Part", remaining_part)
cv2.imshow("Combined-part", combined_part)
cv2.imshow("Comman", comman_part)
revert_img = cv2.bitwise_not(img1)
cv2.imshow("Revert Image", revert_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Contours and Shape Detection

```python
import cv2

img = cv2.imread("vkProfile.jpg", -1)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, thresh = cv2.threshold(gray, 100, 255, cv2.THRESH_BINARY)

# get the coordinates of border
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_NONE)

# draw the coordinates
cv2.drawContours(img, contours, -1, (0, 255, 0), 3)
cv2.imshow("img", img)
cv2.waitKey(0)
cv2.destroyAllWindows()

```

```python
# Shape Detection based on the coordinates
import cv2

img = cv2.imread("shapes.png")
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, thresh = cv2.threshold(gray_img, 120, 255, cv2.THRESH_BINARY)
# GET CONTOURS

contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_NONE)

# draw contours
cv2.drawContours(img, contours, -1, (0, 255, 0), 3)

for contour in contours:
    approx = cv2.approxPolyDP(contour, 0.01 * cv2.arcLength(contour, True), True)
    corner = len(approx)
    chape = "shape"
    if corner == 3:
        shape = "Triangle"
    elif corner == 4:
        shape = "Rectangle"
    elif corner == 5:
        shape = "Pentagon"
    elif corner > 15:
        shape = "Circle"
    else:
        shape = "Unknown"
    color = (0, 255, 0) if shape != "Unknown" else (0, 0, 255)
    print(approx)
    x = approx.ravel()[0]
    y = approx.ravel()[1]
    # print(x, y)
    cv2.putText(img, shape, (x, y), cv2.FONT_HERSHEY_COMPLEX_SMALL, 1, color, 1)
    print(shape)
    
cv2.imshow("Contours", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

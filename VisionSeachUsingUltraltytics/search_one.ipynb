```python
from ultralytics import solutions, YOLO

searcher = solutions.VisualAISearch(
    data="images",
    device="cpu"
)
model = YOLO("yolo26n.pt")

def getClassInfo(image):
    res = model(image)
    searchRes = []
    for cls in res[0].boxes.cls:
        class_id = int(cls)
        class_name = res[0].names[class_id]
        print("Predicted Class Name : ", class_name)
        searchRes.append(searcher(class_name))
    return searchRes
    

def sendFrameToSearch(image):
    classInfo = getClassInfo(image)
    return classInfo
```

```python
import cv2

cam = cv2.VideoCapture(0)
while True:
    success, frame = cam.read()
    if not success:
        break
    cv2.imshow("Video Stream", frame)
    key = cv2.waitKey(1)
    if key == ord('s'):
        res = sendFrameToSearch(frame)
        print("Fetched Records : ", res)
        break
    if key == ord('q'):
        break
cam.release()
cv2.destroyAllWindows()
```
